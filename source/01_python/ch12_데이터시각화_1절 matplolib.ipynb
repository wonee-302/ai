## 1.3 그래프 영역을 나누기 
### 1) subplot() 함수로 서브플롯 추가 

```python
import numpy as np
x= np.linspace(0,9.9,100)
x.shape
```

```python
plt.figure(figsize=(5,5))# 그래프 사이즈가 이번 셀만 
plt.rcParams['figure.figsize']= (5,5)# 그래프 사이즈 셋팅(다음셀에도 적용)
plt.subplot(2,1,1)#2행1열의 subplot에서 1번째 subplot을 다음과 같이 그린다 
plt.plot(x,np.sin(x))
plt.subplot(2,2,3) # 2행2열의 subplot에서 3번째 subplot을 다음과 같이 그린다 
plt.plot(x, np.cos(x))
plt.subplot(224)
plt.plot(x, np.sin(x)*np.cos(x))
plt.show()
```

```python
plt.subplot(221); plt.plot(x, np.sin(x))
plt.subplot(222); plt.plot(x, np.sin(x))
plt.subplot(223); plt.plot(x, np.sin(x))
plt.subplot(2,2,4); plt.plot(x, np.sin(x))
plt.show
```


### 2) subplots() 함수로 서브플롯 생성
- plot함수에서 색상이나 스타일을 조정 : http://wikidocs.net/92085

```python
fig, axes= plt.subplots(nrows=2, ncols=2, figsize=(7,4))
axes[0,0].plot(x, np.sin(x), color='r', linewidth=2 ) # color='r', linewidth=10
axes[0,1].plot(x, np.cos(x), 'b--')#color='b', linestyle='--'(대시라인), :(점선)
axes[1,0].plot(x, np.tanh(x), '#00ff00')# c= '#00ff00'
axes[1,0].plot([0,2,4], [-1,0,1], 'ko-.')#c='k', marker='0', linestyle='-.'
axes[1,1].plot(x, np.sin(x)*np.cos(x), 'r:')#c='k', marker='0', linestyle='-.'
plt.show
```

```python
# 1행 4열 서브플롯(for문)
x= np.linspace(0,9.9,10)
func_list =[np.sin, np.cos, np.tanh, lambda x:np.sin(x)*np.cos(x)]
func_list[3](x)
fmt_list=['ro--', 'bs--', 'go-.', 'r:']
fig, axes = plt.subplots(1,4, figsize=(15,3),sharey=True)# y축 공유
for i, ax in enumerate(axes):
    ax.plot(x, func_list[i](x), fmt_list[i])
axes[2].plot([3,6], [-1, 1], 'kx--')
```

```python
#한글 기본 폰트 설정 방법2
plt.style.use('ggplot')
from matplotlib.font_manager import FontProperties
font_path= 'C:/Windows/Fonts/HMKMMAG.TTF'
#폰트 다운로드 :네이버, 구글, 눈누, 산들
font_prop = FontProperties(fname=font_path, size=15)
```

```python
x=np.linspace(0,10)
y=np.cos(x)
fig, axes =plt.subplots(1,2,figsize=(12,3))
axes[0].plot(x,y,linewidth=5, linestyle='-.', color='purple')#lw=5, ls'-.', c='purple'
#텍스트 추가 
axes[0].text(2,0.5, #텍스트가 위치할 곳의 좌표
            '샘플 텍스트-1', 
            fontsize=30
            )
axes[0].set_title('첫번째 그래프',fontproperties=font_prop)
axes[1].scatter(x,y, marker='^')
#annotate 추가

axes[1].annotate('Sine', xytext=(2,0.75), #글자가 있을 좌표(좌하)
                xy=(5,0.5), #화살표가 가리킬 좌표
                arrowprops=dict(arrowstyle='->', 
                               connectionstyle='angle3')) #angle, angle3, arc, .....

axes[1].set_title('두번째 그래프')
plt.suptitle('메인 제목')
plt.tight_layout()
plt.show()
```

### 4) 축과 눈금 
- https://matplotlib.org/3.1.0/api/axis_api.html  (교안33p) 

```python
fig, axes = plt.subplots(2,2, figsize=(12,5))
#subplot끼리 여백 조정 
plt.subplots_adjust(hspace=0.4, wspace=0.3) #hspace 위아래 여백(1은 서브플롯세로높이)
axes[0,0].scatter(x,y)
axes[0,0].set(title='An example', 
              ylabel='y값',
              xlabel='x값',
             ylim=[-2,2], #ylim은 y축 눈금 범위 
              xlim=[3,6],
              yticks=[-2,-1,0,1,2], # y축 눈금
              yticklabels=['limit_min', -1, '원점', 1, 'limit_max'], #y축 눈금의 라벨
              xticks=[3,4,5,6],
             xticklabels=['삼', '사점',5,'6점'],
             )
axes[0,1].scatter(x,y, c='r', marker='x')
axes[0,1].set_title('두번째', fontsize=15)
axes[0,1].set_xticks(range(0,11,3))
axes[0,1].set_ylim(-1.5, 1.5)

axes[1,0].scatter(x,y)
axes[1,0].spines['top'].set_visible(False)
axes[1,0].spines['bottom'].set_position (('outward', 10)) #10 만큼 바깥으로 이동 

axes[1,1].bar(x,y)
axes[1,1].grid(True)
plt.show()
```

### 5) ★축공유 ★

```python
import numpy as np 
x=np.arange(0,10)
y1=0.5*x**2
y2=-y1
print(x)
print(y1)
print(y2)
```

```python
fig, ax1=plt.subplots(figsize=(10,3))
ax1.plot(x,y1,'g^:')
ax1.set(title='축공유', xlabel='X data', ylabel='y1 data')

ax2= ax1.twinx() # ax1과 x축을 공유하는 ax2 
ax2.plot(x, y2, 'b--')
ax2.set_ylabel('y2 data', color='b')

ax3= ax1.twiny() # ax1과 y축을 공유하는 ax3 
ax3.plot(-x, y1, 'r-')
ax3.set_xlabel('-x data', color='r')
plt.show()
```
