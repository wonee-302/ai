```
모듈/ 패키지나 모듈안의 함수를 가져오는 방법  

import A              - A모듈을 가져옴 
import A as B         - A모듈을 B이름으로 가져옴 
from A import B       - A모듈안의 B함수 (클래스| 변수)를 가져옴 
from A import B as C  - A모듈안의 B 함수를 C 이름으로 가져옴 



from A import B   - A 패키지 안의 B 모듈을 가져옴 
from A.B import C - A 패키지 안의 B 모듈을 안의 C 함수를 가져옴 
from A import B as C - A 패키지 안의 B 모듈을  C 이름으로 가져옴 
from A.B import C as D - A패키지 안의 B 모듈안의 C를 D 이름으로 가져옴 

```
