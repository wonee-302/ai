````python
# 변수의 값을 조회하는 순서 : 지역변수(지역심볼테이블) > 전역변수(전역심볼테이블) > 내장된 이름
g_var = 100 # 전역변수
def func3():
    g_var = 200 # 지역변수
    print('함수안 g_var :', g_var)
func3()
print('함수밖 g_var :', g_var)
````

````python
# 렉시컬 특징 : 변수가 어디에 선언(할당)되었는지에 따라 변수 범위가 정해짐
g_var = 100 # 전역변수
def func3():
    print('함수안 g_var :', g_var)
    g_var = 200 # 지역변수
func3()
print('함수밖 g_var :', g_var)
````

````python
# 렉시컬 특징 : 변수가 어디에 선언(할당)되었는지에 따라 변수 범위가 정해짐
g_var = 100 # 전역변수
def func3():
    global g_var # func3함수내의 g_var는 전역변수를 쓴다
    print('함수안 g_var :', g_var) # 전역변수
    g_var = 200 # 전역변수
func3()
print('함수밖 g_var :', g_var)
````

````python
# args 앞에 * : 매개변수들(인수들)이 튜플로 전달, 튜플 매개변수라고 함 
def add(*args):
    "전달받은 매개변수들의 합을 return. 매개변수가 없을 경우 0을 return"
    #print(args) #(1,2,3)
    tot=0 #누적변수
    for num in args:
        tot +=num #tot=tot+num 
    return tot
````

## 3.7 인수의 언패킹

- 패킹 : 함수 정의시 매개변수 앞에 *붙임(튜플 매개 변수)

````python
def map_template(func,L=[]):
    result =[]
    for item in L:
        result.append(func(item))
    return result 
````
