```python
def commadrop2(row):
    row=row.copy()
    price= row['분양가격(제곱미터당 천원)']
    if pd.isna(price):
        row['제곱미터당분양가격'] =np.nan
    elif price.strip()==''or price=='-': #space거나 -이면 결측치 
        row['제곱미터당분양가격']=np.nan
    elif price.find(',')!= -1: #price에 , 가 포함되었는지 
        row['제곱미터당분양가격']=float(price.replace(',',''))
    else:
        row['제곱미터당분양가격']=float(price)
    return row 

commadrop(df_last.loc[368])
```
